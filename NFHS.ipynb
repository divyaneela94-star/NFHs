# **Dataset Information**

**Source:**  Open Government Data(OGD) Platform India

**Dataset Column Description:**

**Basic Information**

1. District Name – Name of the district

2. State/UT – Name of the state or union territory

3. Number of Households Surveyed – Total households covered in the survey

4. Women / Men Interviewed – Number of women (15–49) and men (15–54) surveyed

**Population & Education Indicators**

1. Population below 15 years – Percentage of children in the population

2. Sex Ratio – Number of females per 1,000 males

3. School Attendance / Literacy – Education level and literacy status of women

**Household Facilities & Living Conditions**

1. Electricity – Households with electricity

2. Drinking Water – Access to improved drinking water

3. Sanitation – Use of improved sanitation facilities

4. Cooking Fuel – Use of clean fuel for cooking

5. Iodized Salt – Households using iodized salt

6. Health Insurance – Coverage under health insurance schemes

**Women & Reproductive Health**

1. Early Marriage – Women married before age 18

2. Teenage Pregnancy – Young women who are mothers or pregnant

3. Family Planning – Use of any and modern contraceptive methods

4. Unmet Need – Women who want family planning but are not using it

**Maternal Health Care**

1. Antenatal Care (ANC) – Check-ups during pregnancy

2. Iron & Folic Acid Intake – Nutrition during pregnancy

3. Postnatal Care – Care received after delivery

4. Institutional Births – Births in health facilities

5. Caesarean Deliveries – Percentage of C-section births

**Child Health & Immunization**

1. Coverage – Children fully vaccinated

2. Specific Vaccines – BCG, Polio, DPT, Measles, Rotavirus

3. Vitamin A – Children receiving Vitamin A supplements

**Child Illness & Nutrition**

1. Diarrhoea & ARI – Prevalence and treatment of childhood illnesses

2. Breastfeeding – Early and exclusive breastfeeding practices

3. Child Nutrition – Stunting, wasting, underweight, overweight

4. Women’s Nutrition & Anaemia

5. BMI Status – Underweight, overweight, or obese women

6. Anaemia – Anaemia among children and women

**Non-Communicable Diseases**

1. Blood Sugar Levels – Diabetes risk among men and women

2. Blood Pressure – Hypertension indicators

**Cancer Screening & Substance Use**

1. Cancer Screening – Cervical, breast, and oral cancer tests

2. Tobacco Use – Tobacco consumption among men and women

3. Alcohol Use – Alcohol consumption among men and women

## Stage 2 – Data Cleaning and Pre-processing



```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.tail()
```

```python
threshold = int(0.7 * len(df))
df = df.dropna(axis=1, thresh=threshold)

```

```python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

```

```python
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python

df.duplicated().sum()


```

```python

df = df.drop_duplicates()
```

```python
Q1 = df[num_cols].quantile(0.25)
Q3 = df[num_cols].quantile(0.75)
IQR = Q3 - Q1

df = df[~((df[num_cols] < (Q1 - 1.5 * IQR)) |
          (df[num_cols] > (Q3 + 1.5 * IQR))).any(axis=1)]
```

```python
df[num_cols].skew()
```

```python
for col in num_cols:
    if df[col].skew() > 1:
        df[col] = np.log1p(df[col])
```

```python

df[num_cols] = df[num_cols].apply(pd.to_numeric, errors='coerce')
```

```python
#  Health risk category using BMI

df['BMI_Risk'] = np.where(
    df['Women (age 15-49 years) whose Body Mass Index (BMI) is below normal (BMI <18.5 kg/m2)21 (%)'] > 30,
    'High Risk',
    'Normal'
)
```

```python
# Determine Child Nutrition Status
df['Child_Nutrition_Status'] = 'Critical'

df.loc[
    df['Children under 5 years who are stunted (height-for-age)18 (%)'] < 25,
    'Child_Nutrition_Status'
] = 'Good'

df.loc[
    (df['Children under 5 years who are stunted (height-for-age)18 (%)'] >= 25) &
    (df['Children under 5 years who are stunted (height-for-age)18 (%)'] < 40),
    'Child_Nutrition_Status'
] = 'Average'

```

```python
#  Identify High Anaemia Risk Districts

df['Anaemia_Risk'] = 'Normal'

df.loc[
    df['All women age 15-49 years who are anaemic22 (%)'] > 50,
    'Anaemia_Risk'
] = 'High Risk'

df.loc[
    (df['All women age 15-49 years who are anaemic22 (%)'] >= 35) &
    (df['All women age 15-49 years who are anaemic22 (%)'] <= 50),
    'Anaemia_Risk'
] = 'Medium Risk'
```

```python
# Classify Districts by Institutional Birth Rate

df['Institutional_Birth_Level'] = 'Low'

df.loc[
    df['Institutional births (in the 5 years before the survey) (%)'] >= 90,
    'Institutional_Birth_Level'
] = 'High'

df.loc[
    (df['Institutional births (in the 5 years before the survey) (%)'] >= 70) &
    (df['Institutional births (in the 5 years before the survey) (%)'] < 90),
    'Institutional_Birth_Level'
] = 'Medium'
```

```python
# Categorize Sanitation Coverage

df['Sanitation_Status'] = 'Poor'

df.loc[
    df['Population living in households that use an improved sanitation facility2 (%)'] >= 80,
    'Sanitation_Status'
] = 'Good'

df.loc[
    (df['Population living in households that use an improved sanitation facility2 (%)'] >= 50) &
    (df['Population living in households that use an improved sanitation facility2 (%)'] < 80),
    'Sanitation_Status'
] = 'Average'
```

```python
#  Health Insurance Coverage Level

df['Insurance_Coverage_Level'] = 'Low'

df.loc[
    df['Households with any usual member covered under a health insurance/financing scheme (%)'] >= 60,
    'Insurance_Coverage_Level'
] = 'High'

df.loc[
    (df['Households with any usual member covered under a health insurance/financing scheme (%)'] >= 30) &
    (df['Households with any usual member covered under a health insurance/financing scheme (%)'] < 60),
    'Insurance_Coverage_Level'
] = 'Medium'
```

```python
# Blood Pressure Risk Classification (Women)

df['BP_Risk_Women'] = 'Normal'

df.loc[
    df['Women age 15 years and above wih Elevated blood pressure (Systolic ≥140 mm of Hg and/or Diastolic ≥90 mm of Hg) or taking medicine to control blood pressure (%)'] > 30,
    'BP_Risk_Women'
] = 'High Risk'

df.loc[
    (df['Women age 15 years and above wih Elevated blood pressure (Systolic ≥140 mm of Hg and/or Diastolic ≥90 mm of Hg) or taking medicine to control blood pressure (%)'] >= 15) &
    (df['Women age 15 years and above wih Elevated blood pressure (Systolic ≥140 mm of Hg and/or Diastolic ≥90 mm of Hg) or taking medicine to control blood pressure (%)'] <= 30),
    'BP_Risk_Women'
] = 'Moderate Risk'
```

```python
value = 85
if value >= 90:
    result = "Excellent"
elif value >= 70:
    result = "Good"
else:
    result = "Needs Improvement"

print(result)
```

```python
# Insurance Coverage vs Anaemia Risk vs Child Nutrition

pivot_table = pd.pivot_table(
    df,
    index='Insurance_Coverage_Level',
    columns='Anaemia_Risk',
    values='Children under 5 years who are stunted (height-for-age)18 (%)',
    aggfunc='mean'
)

pivot_table.plot(kind='bar', figsize=(8,5))
plt.title('Insurance Coverage vs Anaemia Risk vs Child Stunting')
plt.xlabel('Insurance Coverage Level')
plt.ylabel('Average Stunting (%)')
plt.show()
```

1. The pivot table calculates the average percentage of stunted children by grouping districts according to insurance coverage level and anaemia risk category.

2. * X-axis- Shows Health Insurance Coverage Level of districts (Low, Medium, High).
   * Y-axis- Shows the average percentage of children under 5 who are stunted.

3. Districts with higher insurance coverage and lower anaemia risk generally show lower child stunting, indicating better health outcomes.
